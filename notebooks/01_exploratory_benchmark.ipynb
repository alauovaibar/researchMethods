## 2. Inspect Sample Benchmark Results

```python
results_path = "../results/benchmark_sample_results.json"
if os.path.exists(results_path):
    with open(results_path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    print(f"Loaded {len(data.get('runs', []))} run(s) from {results_path}")
    for run in data.get('runs', []):
        cfg = run['configuration']
        metrics = run['mean_metrics']
        print(f"top_k={cfg['top_k']} | chunk_size={cfg['chunk_size']} -> Hallucination Rate: {metrics['hallucination_rate']:.4f}, Faithfulness: {metrics['faithfulness_score']:.4f}")
else:
    print("Results file not found. Run benchmark.py first.")
```
